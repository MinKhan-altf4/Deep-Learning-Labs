```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())
```

```python
import numpy as np
import torch

torch.manual_seed(0)
np.random.seed(0)

# ---------- Câu 1: reshape ----------
print("=== Câu 1: reshape ===")
X = torch.arange(24)
print("X:", X)
print("4 x 6:\n", X.reshape(4, 6))
print("2 x 12:\n", X.reshape(2, 12))
print("2 x 3 x 4:\n", X.reshape(2, 3, 4))

# ---------- Câu 2: phép toán trên tensor ----------
print("\n=== Câu 2: phép toán trên tensor ===")
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("A:\n", A)
print("B:\n", B)
print("A + B:\n", A + B)
print("A * B:\n", A * B)
print("A / B:\n", A / B)
print("Trung bình theo cột:", A.mean(dim=0))
print("Độ lệch chuẩn theo cột:", A.std(dim=0))

# ---------- Câu 3: autograd ----------
print("\n=== Câu 3: autograd ===")
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
y.backward()
print("dy/dx tại x = 2 (autograd):", x.grad.item())
print("dy/dx tại x = 2 (thủ công, 4x + 5):", 4 * 2 + 5)

# ---------- Câu 4: huấn luyện đa thức y = 2x^2 - 3x + 5 ----------
print("\n=== Câu 4: học hệ số của y = 2x^2 - 3x + 5 ===")
polynomial = np.poly1d([2, -3, 5])

N = 20
Xn = np.random.randn(N, 1) * 5
Yn = polynomial(Xn)

XX = np.hstack([Xn * Xn, Xn, np.ones_like(Xn)])

w = torch.randn(3, 1, requires_grad=True)
x_t = torch.tensor(XX, dtype=torch.float32)
y_t = torch.tensor(Yn, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.05)

print("Hệ số ban đầu:", w.detach().flatten().tolist())

for epoch in range(5000):
    optimizer.zero_grad()
    y_pred = x_t @ w
    mse = torch.mean(torch.square(y_t - y_pred))
    mse.backward()
    optimizer.step()

print("Hệ số học được [a, b, c]:", w.detach().flatten().tolist())
print("Hệ số thực          :", [2, -3, 5])
print("MSE cuối:", mse.item())
```
